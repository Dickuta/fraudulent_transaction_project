```python
# 2. Remove rows where timestamp is missing
df = df.dropna(subset=['timestamp'])
```

```python
# 3. Create missingness indicators for numerical variables
df['amount_missing'] = df['amount_usd'].isna().astype(int)
df['fee_missing'] = df['fee'].isna().astype(int)
df['device_trust_missing'] = df['device_trust_score'].isna().astype(int)
```

```python
# 4. Median imputation for numerical variables
df['amount_usd'] = df['amount_usd'].fillna(df['amount_usd'].median())
df['fee'] = df['fee'].fillna(df['fee'].median())
df['device_trust_score'] = df['device_trust_score'].fillna(
    df['device_trust_score'].median()
)
```

```python
# 5. Fill categorical/text variables
df['ip_address'] = df['ip_address'].fillna('Unknown')
df['ip_country'] = df['ip_country'].fillna('Unknown')
df['kyc_tier'] = df['kyc_tier'].fillna('Unknown')
```
